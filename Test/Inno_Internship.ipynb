```python
import pandas as pd

# -----------------------------------------
# 1. Filter FAMILY category
# -----------------------------------------
family_df = merged[merged["Category"].astype(str).str.lower() == "family"]

# -----------------------------------------
# 2. Clean Reviews column
# -----------------------------------------
family_df["Reviews"] = (
    family_df["Reviews"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

family_df["Reviews"] = pd.to_numeric(family_df["Reviews"], errors="coerce")

# -----------------------------------------
# 3. Calculate IQR for Reviews
# -----------------------------------------
Q1 = family_df["Reviews"].quantile(0.25)
Q3 = family_df["Reviews"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# -----------------------------------------
# 4. Count outliers
# -----------------------------------------
outliers = family_df[(family_df["Reviews"] < lower_limit) | (family_df["Reviews"] > upper_limit)]

num_outliers = len(outliers)
total = len(family_df)

percentage_outliers = (num_outliers / total) * 100

# -----------------------------------------
# 5. Print results
# -----------------------------------------
print(f"Total FAMILY apps: {total}")
print(f"Number of outliers in Reviews: {num_outliers}")
print(f"Percentage of outliers: {percentage_outliers:.2f}%")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# ---------------------------------------
# 1. Clean the relevant columns
# ---------------------------------------
merged["Type"] = merged["Type"].astype(str).str.strip().str.lower()
merged["Sentiment"] = merged["Sentiment"].astype(str).str.strip().str.lower()

# Keep only valid sentiment categories
merged = merged[merged["Sentiment"].isin(["positive", "negative", "neutral"])]

# ---------------------------------------
# 2. Create contingency table
# ---------------------------------------
contingency_table = pd.crosstab(merged["Type"], merged["Sentiment"])
print("Contingency Table:")
print(contingency_table)

# ---------------------------------------
# 3. Perform Chi-Square Test
# ---------------------------------------
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Test Results:")
print(f"Chi-square statistic = {chi2}")
print(f"Degrees of freedom = {dof}")
print(f"P-value = {p_value}")
print("\nExpected Frequencies:")
print(expected)

# ---------------------------------------
# 4. Conclusion at α = 0.05
# ---------------------------------------
alpha = 0.05
if p_value < alpha:
    print("\n📌 Conclusion: Reject the null hypothesis.")
    print("➡ There IS a significant association between App Type and User Sentiment.")
else:
    print("\n📌 Conclusion: Fail to reject the null hypothesis.")
    print("➡ There is NO significant association between App Type and User Sentiment.")
```

```python
import pandas as pd
import re
from collections import Counter
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop = set(stopwords.words('english'))

# Filter only negative reviews
neg = merged[merged["Sentiment"].str.lower() == "negative"]["Translated_Review"].dropna()

# Text preprocessing
cleaned_words = []

for text in neg:
    # lowercase
    text = text.lower()
    # remove special characters
    text = re.sub(r'[^a-z\s]', '', text)
    # split into words
    words = text.split()
    # remove stopwords
    words = [w for w in words if w not in stop]
    cleaned_words.extend(words)

# Find most common word
most_common_word, freq = Counter(cleaned_words).most_common(1)[0]

print("Most frequent word:", most_common_word)
print("Frequency:", freq)

```

```python
import pandas as pd
import numpy as np

# --- 1. Convert Size column to string first ---
merged["Size"] = merged["Size"].astype(str)

# --- 2. Remove rows where size is "Varies with device" ---
filtered = merged[merged["Size"].str.lower() != "varies with device"]

# --- 3. Function to convert size values to MB ---
def convert_size(x):
    x = str(x).strip().lower()

    if x.endswith("m"):              # MB
        return float(x.replace("m", ""))
    elif x.endswith("k"):            # KB → MB
        return float(x.replace("k", "")) / 1024
    else:
        return np.nan                # Unknown numbers → NaN

# --- 4. Apply conversion ---
filtered["Size_MB"] = filtered["Size"].apply(convert_size)

# --- 5. Drop missing converted values ---
filtered = filtered.dropna(subset=["Size_MB"])

# --- 6. Compute the median size ---
median_size = filtered["Size_MB"].median()

print("Median App Size (in MB):", median_size)
```
